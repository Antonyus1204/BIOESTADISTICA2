Un valor ausente **no equivale a cero**. Representa información que no fue observada, registrada o recuperada.

En un análisis real, la decisión sobre un dato ausente depende de la proporción de ausencia, el mecanismo de ausencia, la variable afectada, el tamaño muestral, el contexto biológico y el impacto potencial sobre el análisis.

En este taller se realizará una imputación sencilla **solo con fines didácticos** y siempre documentada.

### Decisión didáctica de imputación

Para este ejercicio:

- `biomasa_g`: se imputará con la **mediana del grupo experimental**.
- `clorofila_SPAD`: se imputará con la **mediana del grupo experimental**.

Esta estrategia es más conservadora que usar una media global cuando existen diferencias entre grupos. En una investigación real debe justificarse formalmente.

````python
# ============================================================
# CELDA 9. CREACIÓN DE COPIA DE TRABAJO
# ============================================================
# Nunca modificamos directamente el DataFrame original.
# Creamos una copia para conservar trazabilidad.
df_limpio = df.copy()
# Confirmamos que la copia fue creada.
print("Copia de trabajo creada:", df_limpio.shape)
````

````python
# ============================================================
# CELDA 10. IMPUTACIÓN DOCUMENTADA POR MEDIANA DEL GRUPO
# ============================================================
# Calculamos la mediana de biomasa dentro de cada grupo.
mediana_biomasa_grupo = df_limpio.groupby("grupo")["biomasa_g"].transform("median")
# Reemplazamos solo los valores ausentes de biomasa.
df_limpio["biomasa_g"] = df_limpio["biomasa_g"].fillna(mediana_biomasa_grupo)
# Calculamos la mediana de clorofila dentro de cada grupo.
mediana_clorofila_grupo = df_limpio.groupby("grupo")["clorofila_SPAD"].transform("median")
# Reemplazamos solo los valores ausentes de clorofila.
df_limpio["clorofila_SPAD"] = df_limpio["clorofila_SPAD"].fillna(mediana_clorofila_grupo)
# Verificamos que ya no existan datos ausentes en esas variables.
print(df_limpio[["biomasa_g", "clorofila_SPAD"]].isna().sum())
````

## 7. Control de calidad: registros duplicados

Un duplicado puede inflar artificialmente el tamaño muestral, alterar medias y proporciones, reducir de forma artificial la variabilidad y aumentar falsamente la precisión.

Debe distinguirse entre un **duplicado real** y dos observaciones legítimamente iguales.

````python
# ============================================================
# CELDA 11. DETECCIÓN DE DUPLICADOS
# ============================================================
# duplicated() identifica filas repetidas.
duplicados_booleanos = df_limpio.duplicated()
# sum() cuenta el número total de duplicados exactos.
n_duplicados = duplicados_booleanos.sum()
# Mostramos el total.
print("Número de duplicados exactos:", n_duplicados)
# Mostramos los registros duplicados para revisarlos.
display(df_limpio[df_limpio.duplicated(keep=False)])
````

## 8. Control de calidad: valores atípicos

Un valor atípico es una observación extrema respecto a la distribución.

**Importante:** un outlier no es automáticamente un error. Puede representar error de medición, error de transcripción, unidad incorrecta, evento biológico real o subgrupo no reconocido.

Usaremos el criterio del rango intercuartílico (**IQR**) como herramienta de detección, no como regla automática de exclusión.

````python
# ============================================================
# CELDA 13. FUNCIÓN PARA DETECTAR OUTLIERS POR IQR
# ============================================================
# Definimos una función reutilizable.
def detectar_outliers_iqr(serie):
    # Calculamos el primer cuartil.
    q1 = serie.quantile(0.25)
    # Calculamos el tercer cuartil.
    q3 = serie.quantile(0.75)
    # Calculamos el rango intercuartílico.
    iqr = q3 - q1
    # Definimos el límite inferior.
    limite_inferior = q1 - 1.5 * iqr
    # Definimos el límite superior.
    limite_superior = q3 + 1.5 * iqr
    # Identificamos observaciones fuera de los límites.
    mascara = (serie < limite_inferior) | (serie > limite_superior)
    # Devolvemos resultados.
    return mascara, q1, q3, iqr, limite_inferior, limite_superior
````

````python
# ============================================================
# CELDA 14. DETECCIÓN DE OUTLIERS EN LONGITUD
# ============================================================
# Aplicamos la función a longitud_cm.
mascara_longitud, q1_l, q3_l, iqr_l, li_l, ls_l = detectar_outliers_iqr(df_limpio["longitud_cm"])
# Mostramos los límites calculados.
print("Q1:", round(q1_l, 2))
print("Q3:", round(q3_l, 2))
print("IQR:", round(iqr_l, 2))
print("Límite inferior:", round(li_l, 2))
print("Límite superior:", round(ls_l, 2))
# Mostramos las observaciones potencialmente atípicas.
display(df_limpio.loc[mascara_longitud])
````

````python
# ============================================================
# CELDA 15. REVISIÓN DE PLAUSIBILIDAD BIOLÓGICA
# ============================================================
# Seleccionamos valores extremos de longitud para revisión.
revision_longitud = df_limpio.loc[mascara_longitud, ["id_planta", "grupo", "longitud_cm", "biomasa_g", "clorofila_SPAD"]]
# Mostramos la tabla de revisión.
display(revision_longitud)
# En este taller NO eliminaremos automáticamente estos valores.
print("Decisión didáctica: conservar el valor y marcarlo para revisión.")
````

````python
# ============================================================
# CELDA 16. DETECCIÓN DE VALORES SOSPECHOSOS DE pH
# ============================================================
# Definimos un rango docente plausible para este experimento.
# Este rango es solo un criterio de práctica, no una regla universal.
ph_min = 5.5
ph_max = 7.5
# Identificamos valores fuera del rango definido.
ph_sospechoso = (df_limpio["pH_sustrato"] < ph_min) | (df_limpio["pH_sustrato"] > ph_max)
# Mostramos las observaciones que requieren revisión.
display(df_limpio.loc[ph_sospechoso, ["id_planta", "grupo", "pH_sustrato"]])
# Recordamos que el criterio debe ser contextual.
print("Estos valores requieren revisión de unidad, instrumento y registro.")
````

## 9. Estadística descriptiva

La estadística descriptiva resume el comportamiento de los datos sin realizar todavía inferencia.

### Tendencia central
- **Media:** sensible a valores extremos.
- **Mediana:** robusta frente a asimetría y outliers.
- **Moda:** valor o categoría más frecuente.

### Dispersión
- **Rango:** máximo − mínimo.
- **Varianza:** dispersión cuadrática respecto a la media.
- **Desviación estándar:** dispersión en la misma unidad de la variable.
- **IQR:** Q3 − Q1, resume el 50 % central.

````python
# ============================================================
# CELDA 17. RESUMEN DESCRIPTIVO GENERAL
# ============================================================
# describe() genera un resumen descriptivo de variables numéricas.
resumen_general = df_limpio.describe()
# Mostramos el resumen.
display(resumen_general)
````

````python
# ============================================================
# CELDA 18. CÁLCULO MANUAL DE MEDIDAS PARA LONGITUD
# ============================================================
# Seleccionamos la variable longitud_cm.
x = df_limpio["longitud_cm"]
# Calculamos la media.
media = x.mean()
# Calculamos la mediana.
mediana = x.median()
# Calculamos la moda.
moda = x.mode().tolist()
# Calculamos el rango.
rango = x.max() - x.min()
# Calculamos la varianza muestral.
varianza = x.var(ddof=1)
# Calculamos la desviación estándar muestral.
desviacion = x.std(ddof=1)
# Calculamos Q1.
q1 = x.quantile(0.25)
# Calculamos Q3.
q3 = x.quantile(0.75)
# Calculamos el IQR.
iqr = q3 - q1
# Mostramos los resultados.
print("Media:", round(media, 3))
print("Mediana:", round(mediana, 3))
print("Moda:", moda)
print("Rango:", round(rango, 3))
print("Varianza muestral:", round(varianza, 3))
print("Desviación estándar muestral:", round(desviacion, 3))
print("IQR:", round(iqr, 3))
````

### Interpretación científica

Compare media y mediana:

- Si son similares, la distribución puede ser aproximadamente simétrica.
- Si difieren notablemente, revise asimetría y valores extremos.
- Una desviación estándar elevada debe interpretarse según contexto biológico; no implica automáticamente mala calidad.

````python
# ============================================================
# CELDA 19. FRECUENCIAS DE LA VARIABLE CATEGÓRICA
# ============================================================
# value_counts() cuenta cuántas observaciones hay por grupo.
frecuencias_grupo = df_limpio["grupo"].value_counts()
# Mostramos frecuencias absolutas.
print("Frecuencia por grupo:")
print(frecuencias_grupo)
# Calculamos porcentajes.
porcentajes_grupo = df_limpio["grupo"].value_counts(normalize=True).mul(100).round(1)
# Mostramos porcentajes.
print()
print("Porcentaje por grupo:")
print(porcentajes_grupo)
````

## 10. Resumen por grupos con `groupby()`

En ciencias biológicas, las medidas descriptivas suelen interpretarse por condición experimental.

El objetivo aquí no es todavía afirmar significancia estadística, sino comparar tamaño muestral, media, mediana, desviación estándar e IQR.

````python
# ============================================================
# CELDA 20. RESUMEN POR GRUPO
# ============================================================
# Agrupamos por condición experimental y resumimos longitud.
resumen_grupo = df_limpio.groupby("grupo")["longitud_cm"].agg(["count", "mean", "median", "std", "min", "max"]).round(2)
# Mostramos la tabla.
display(resumen_grupo)
````

````python
# ============================================================
# CELDA 21. IQR POR GRUPO
# ============================================================
# Definimos una función para calcular IQR.
def calcular_iqr(serie):
    # Calculamos Q1.
    q1 = serie.quantile(0.25)
    # Calculamos Q3.
    q3 = serie.quantile(0.75)
    # Retornamos la diferencia.
    return q3 - q1
# Calculamos IQR de longitud para cada grupo.
iqr_por_grupo = df_limpio.groupby("grupo")["longitud_cm"].apply(calcular_iqr).round(2)
# Mostramos el resultado.
print(iqr_por_grupo)
````
