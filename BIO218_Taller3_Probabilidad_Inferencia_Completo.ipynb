- La distribución estadística debe relacionarse con el **proceso generador de datos**.

# PARTE I. PROBABILIDAD Y DISTRIBUCIONES

## 6. Probabilidad simple

Una probabilidad toma valores entre 0 y 1:

\[
0 \le P(A) \le 1
\]

En un contexto empírico sencillo:

\[
P(A)=\frac{\text{número de resultados favorables}}{\text{número total de observaciones}}
\]

Usaremos la variable `germino`.

```python
# ============================================================
# CELDA 4. PROBABILIDAD EMPÍRICA DE GERMINACIÓN
# ============================================================
# Contamos cuántas plantas germinaron.
n_germinaron = df["germino"].sum()

# Contamos el número total de observaciones.
n_total = len(df)

# Calculamos la proporción observada de germinación.
p_germinacion = n_germinaron / n_total

# Mostramos el resultado.
print("Plantas germinadas:", n_germinaron)
print("Total de plantas:", n_total)
print("P(germinación) estimada =", round(p_germinacion, 3))
```

### Interpretación

La proporción observada es una **estimación muestral** de la probabilidad de germinación bajo las condiciones representadas en el dataset.

No debe interpretarse automáticamente como una constante universal.

## 7. Probabilidad condicional

La probabilidad condicional se expresa como:

\[
P(A|B)=\frac{P(A\cap B)}{P(B)}
\]

Aquí preguntaremos:

> **¿Cuál es la probabilidad observada de germinar entre las plantas del Tratamiento_A?**

```python
# ============================================================
# CELDA 5. PROBABILIDAD CONDICIONAL
# ============================================================
# Seleccionamos únicamente las plantas del Tratamiento_A.
df_A = df[df["grupo"] == "Tratamiento_A"]

# Calculamos la proporción de germinación dentro de ese grupo.
p_germina_dado_A = df_A["germino"].mean()

# Mostramos el resultado.
print("P(germinar | Tratamiento_A) =", round(p_germina_dado_A, 3))

# Calculamos también el Control para comparación descriptiva.
df_control = df[df["grupo"] == "Control"]

# Estimamos P(germinar | Control).
p_germina_dado_control = df_control["germino"].mean()

# Mostramos la comparación.
print("P(germinar | Control) =", round(p_germina_dado_control, 3))
```

### Pregunta de reflexión

¿La diferencia entre esas dos probabilidades observadas implica causalidad?

**Respuesta:** no. La causalidad requiere considerar diseño, aleatorización, control, independencia y otras fuentes de sesgo.

## 8. Distribución normal

La distribución normal se utiliza para modelar variables continuas aproximadamente simétricas alrededor de una media.

Parámetros:

- `μ`: media.
- `σ`: desviación estándar.

Usaremos una simulación para visualizar el concepto.

```python
# ============================================================
# CELDA 6. SIMULACIÓN DE UNA DISTRIBUCIÓN NORMAL
# ============================================================
# Creamos un generador aleatorio reproducible.
rng = np.random.default_rng(218)

# Simulamos 1000 longitudes con media 12 cm y DE 1.5 cm.
longitudes_simuladas = rng.normal(
    loc=12,
    scale=1.5,
    size=1000
)

# Calculamos media y desviación estándar de la simulación.
print("Media simulada:", round(longitudes_simuladas.mean(), 3))
print("DE simulada:", round(longitudes_simuladas.std(ddof=1), 3))

# Construimos un histograma.
plt.figure(figsize=(8, 5))
plt.hist(longitudes_simuladas, bins=25, edgecolor="black", alpha=0.8)

# Añadimos etiquetas científicas.
plt.title("Distribución normal simulada de longitud")
plt.xlabel("Longitud (cm)")
plt.ylabel("Frecuencia")

# Ajustamos y mostramos.
plt.tight_layout()
plt.show()
```

## 9. Distribución binomial

La binomial modela el número de éxitos en `n` ensayos cuando:

- hay dos resultados posibles;
- el número de ensayos es fijo;
- los ensayos son independientes;
- la probabilidad `p` permanece razonablemente estable.

Ejemplo: número de semillas germinadas de un total de 20.

```python
# ============================================================
# CELDA 7. SIMULACIÓN BINOMIAL
# ============================================================
# Simulamos 100 experimentos.
# En cada experimento se observan 20 semillas.
# La probabilidad de germinación se fija en 0.70.
germinadas_sim = rng.binomial(
    n=20,
    p=0.70,
    size=100
)

# Mostramos las primeras simulaciones.
print("Primeros 10 experimentos:", germinadas_sim[:10])

# Calculamos el promedio de semillas germinadas.
print("Media observada:", round(germinadas_sim.mean(), 2))

# Visualizamos la distribución de conteos.
plt.figure(figsize=(8, 5))
plt.hist(germinadas_sim, bins=np.arange(germinadas_sim.min()-0.5, germinadas_sim.max()+1.5, 1), edgecolor="black", alpha=0.8)

# Añadimos etiquetas.
plt.title("Distribución binomial simulada")
plt.xlabel("Número de semillas germinadas de 20")
plt.ylabel("Frecuencia")

# Mostramos la figura.
plt.tight_layout()
plt.show()
```

## 10. Distribución de Poisson

La distribución de Poisson modela conteos de eventos que ocurren en una unidad definida de:

- tiempo;
- espacio;
- volumen;
- longitud genómica;
- campo microscópico.

Su parámetro `λ` representa la tasa media esperada.

```python
# ============================================================
# CELDA 8. SIMULACIÓN POISSON
# ============================================================
# Simulamos 200 observaciones con una tasa media de 3 eventos.
eventos_poisson = rng.poisson(
    lam=3,
    size=200
)

# Mostramos la media observada.
print("Media del conteo simulado:", round(eventos_poisson.mean(), 3))

# Construimos la distribución.
plt.figure(figsize=(8, 5))
plt.hist(eventos_poisson, bins=np.arange(eventos_poisson.min()-0.5, eventos_poisson.max()+1.5, 1), edgecolor="black", alpha=0.8)

# Añadimos etiquetas.
plt.title("Distribución Poisson simulada")
plt.xlabel("Número de eventos por unidad")
plt.ylabel("Frecuencia")

# Mostramos.
plt.tight_layout()
plt.show()
```

## 11. Variabilidad muestral

Si tomamos distintas muestras de una misma población, sus medias no serán idénticas.

La **distribución muestral de la media** describe cómo varía la media entre muestras repetidas.

A medida que aumenta `n`, el error estándar de la media suele disminuir:

\[
SE(\bar{x})=\frac{s}{\sqrt{n}}
\]

```python
# ============================================================
# CELDA 9. CREACIÓN DE UNA POBLACIÓN SIMULADA
# ============================================================
# Simulamos una población grande de longitudes.
poblacion = rng.normal(
    loc=12,
    scale=1.5,
    size=10000
)

# Calculamos parámetros aproximados de la población simulada.
print("Media poblacional simulada:", round(poblacion.mean(), 3))
print("DE poblacional simulada:", round(poblacion.std(ddof=1), 3))
```

```python
# ============================================================
# CELDA 10. DISTRIBUCIÓN MUESTRAL CON n = 5
# ============================================================
# Creamos una lista para almacenar medias muestrales.
medias_n5 = []

# Repetimos 1000 veces el proceso de muestreo.
for _ in range(1000):
    # Extraemos una muestra aleatoria de 5 observaciones.
    muestra = rng.choice(poblacion, size=5, replace=False)

    # Calculamos y almacenamos su media.
    medias_n5.append(muestra.mean())

# Convertimos la lista a un array.
medias_n5 = np.array(medias_n5)

# Mostramos la desviación estándar de las medias muestrales.
print("DE de medias muestrales con n=5:", round(medias_n5.std(ddof=1), 3))

# Visualizamos la distribución muestral.
plt.figure(figsize=(8, 5))
plt.hist(medias_n5, bins=25, edgecolor="black", alpha=0.8)

# Añadimos etiquetas.
plt.title("Distribución muestral de la media: n = 5")
plt.xlabel("Media muestral de longitud")
plt.ylabel("Frecuencia")

# Mostramos.
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CELDA 11. DISTRIBUCIÓN MUESTRAL CON n = 30
# ============================================================
# Creamos una lista para almacenar medias muestrales.
medias_n30 = []

# Repetimos el proceso 1000 veces.
for _ in range(1000):
    # Extraemos una muestra de 30 observaciones.
    muestra = rng.choice(poblacion, size=30, replace=False)

    # Calculamos la media.
    medias_n30.append(muestra.mean())

# Convertimos a array.
medias_n30 = np.array(medias_n30)

# Comparamos la variabilidad.
print("DE de medias muestrales con n=30:", round(medias_n30.std(ddof=1), 3))

# Visualizamos.
plt.figure(figsize=(8, 5))
plt.hist(medias_n30, bins=25, edgecolor="black", alpha=0.8)

# Añadimos etiquetas.
plt.title("Distribución muestral de la media: n = 30")
plt.xlabel("Media muestral de longitud")
plt.ylabel("Frecuencia")

# Mostramos.
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CELDA 12. COMPARACIÓN NUMÉRICA DEL ERROR MUESTRAL
# ============================================================
# Calculamos la desviación estándar de las medias para n=5.
se_empirico_n5 = medias_n5.std(ddof=1)

# Calculamos la desviación estándar de las medias para n=30.
se_empirico_n30 = medias_n30.std(ddof=1)

# Mostramos ambos valores.
print("Error muestral empírico aproximado, n=5 :", round(se_empirico_n5, 3))
print("Error muestral empírico aproximado, n=30:", round(se_empirico_n30, 3))

# Interpretación:
# Un tamaño muestral mayor suele producir una estimación más precisa.
```

# PARTE II. INFERENCIA ESTADÍSTICA

## 12. Hipótesis estadísticas

Pregunta biológica:

> **¿El tratamiento modifica la longitud media de las plántulas?**

Para Control vs Tratamiento_A:

\[
H_0: \mu_C = \mu_A
\]

\[
H_1: \mu_C \ne \mu_A
\]

Usaremos una prueba **bilateral** porque la pregunta plantea diferencia en cualquiera de las dos direcciones.

La dirección de la hipótesis debe definirse **antes de observar el resultado**, no después.

## 13. Nivel de significancia α

`α` es el umbral de error tipo I fijado antes del análisis.

En este taller:

\[
\alpha = 0.05
\]

- Error tipo I: rechazar H₀ cuando H₀ es verdadera.
- Error tipo II: no rechazar H₀ cuando existe un efecto real.

`α = 0.05` **no significa** “95 % de probabilidad de que el resultado sea verdadero”.

## 14. Interpretación correcta del p-valor

El p-valor cuantifica qué tan compatibles son los datos observados con H₀, bajo el modelo y los supuestos de la prueba.

Regla operativa:

- `p ≤ α` → rechazar H₀.
- `p > α` → no rechazar H₀.

Nunca escribir:

> “p es la probabilidad de que H₀ sea verdadera”.

Tampoco debe confundirse un p-valor pequeño con un efecto biológicamente importante.

## 16. Supuesto 1: independencia

La independencia **no se demuestra con Shapiro ni con Levene**.

Debe justificarse con el diseño:

- cada planta representa una unidad experimental independiente;
- no se reutiliza una misma unidad como si fueran individuos diferentes;
- las réplicas técnicas no se confunden con réplicas biológicas;
- la asignación y el muestreo deben evitar pseudorreplicación.

## 17. Supuesto 2: normalidad

En pruebas paramétricas, el supuesto se refiere especialmente a la distribución de errores/residuos. En un ejercicio introductorio también podemos explorar los datos por grupo.

Utilizaremos:

- histograma;
- Q-Q plot;
- Shapiro–Wilk.

**Advertencia:** una prueba de normalidad no debe interpretarse de forma aislada.

```python
# ============================================================
# CELDA 15. SHAPIRO-WILK POR GRUPO
# ============================================================
# Recorremos cada grupo experimental.
for grupo in df["grupo"].unique():

    # Seleccionamos la longitud del grupo actual.
    valores = df.loc[df["grupo"] == grupo, "longitud_cm"]

    # Aplicamos la prueba de Shapiro-Wilk.
    W, p = stats.shapiro(valores)

    # Mostramos resultados.
    print(f"{grupo}: W = {W:.4f}, p = {p:.4f}")

# Interpretación:
# p > 0.05 indica que no encontramos evidencia suficiente
# para rechazar normalidad con esta prueba.
# Esto NO demuestra que la distribución sea perfectamente normal.
```

```python
# ============================================================
# CELDA 16. Q-Q PLOT DEL GRUPO CONTROL
# ============================================================
# Seleccionamos el grupo Control.
x_control = df.loc[df["grupo"] == "Control", "longitud_cm"]

# Creamos la figura.
plt.figure(figsize=(6, 6))

# Generamos un gráfico Q-Q frente a una distribución normal.
stats.probplot(x_control, dist="norm", plot=plt)

# Añadimos título.
plt.title("Q-Q plot: longitud en grupo Control")

# Mostramos.
plt.tight_layout()
plt.show()
```

## 18. Supuesto 3: homogeneidad de varianzas

La prueba de Levene evalúa si la dispersión de los grupos es compatible con varianzas similares.

- `p > 0.05`: no hay evidencia suficiente para rechazar homogeneidad.
- `p ≤ 0.05`: existe evidencia de heterogeneidad.

Si las varianzas son desiguales en una comparación de dos grupos, puede utilizarse la **t de Welch** en lugar de asumir varianzas iguales.

```python
# ============================================================
# CELDA 17. PRUEBA DE LEVENE
# ============================================================
# Extraemos las longitudes de cada grupo.
x_control = df.loc[df["grupo"] == "Control", "longitud_cm"]
x_A = df.loc[df["grupo"] == "Tratamiento_A", "longitud_cm"]
x_B = df.loc[df["grupo"] == "Tratamiento_B", "longitud_cm"]

# Aplicamos Levene a los tres grupos.
estadistico_levene, p_levene = stats.levene(
    x_control,
    x_A,
    x_B
)

# Mostramos resultados.
print("Levene:")
print("Estadístico =", round(estadistico_levene, 4))
print("p-valor     =", round(p_levene, 4))
```

# PARTE III. COMPARACIÓN DE DOS GRUPOS

## 19. t de Student para dos grupos independientes

Se utiliza cuando:

- la variable dependiente es cuantitativa;
- los grupos son independientes;
- la distribución es razonablemente compatible con normalidad;
- la homogeneidad de varianzas es razonable, o se utiliza Welch.

Compararemos:

**Control vs Tratamiento_A**

```python
# ============================================================
# CELDA 18. t DE STUDENT: CONTROL VS TRATAMIENTO_A
# ============================================================
# Realizamos una prueba bilateral para dos grupos independientes.
t_stat, p_t = stats.ttest_ind(
    x_control,
    x_A,
    equal_var=True,
    alternative="two-sided"
)

# Mostramos el estadístico y p-valor.
print("t =", round(t_stat, 4))
print("p =", format(p_t, ".6g"))

# Regla de decisión docente.
alpha = 0.05

# Interpretamos la decisión estadística.
if p_t <= alpha:
    print("Decisión: rechazar H0.")
else:
    print("Decisión: no rechazar H0.")
```

### No basta con el p-valor

Debemos estimar:

- diferencia de medias;
- intervalo de confianza;
- tamaño del efecto.

```python
# ============================================================
# CELDA 19. DIFERENCIA DE MEDIAS E IC95%
# ============================================================
# Calculamos las medias.
media_control = x_control.mean()
media_A = x_A.mean()

# Calculamos la diferencia Control - Tratamiento_A.
dif_medias = media_control - media_A

# Calculamos tamaños muestrales.
n1 = len(x_control)
n2 = len(x_A)

# Calculamos varianzas muestrales.
s1_2 = x_control.var(ddof=1)
s2_2 = x_A.var(ddof=1)

# Calculamos la varianza combinada.
var_pooled = (
    ((n1 - 1) * s1_2 + (n2 - 1) * s2_2)
    / (n1 + n2 - 2)
)

# Calculamos el error estándar de la diferencia.
se_dif = np.sqrt(
    var_pooled * (1/n1 + 1/n2)
)

# Calculamos grados de libertad.
gl = n1 + n2 - 2

# Obtenemos el valor crítico t para un IC95%.
t_crit = stats.t.ppf(0.975, df=gl)

# Calculamos límites.
li = dif_medias - t_crit * se_dif
ls = dif_medias + t_crit * se_dif

# Mostramos resultados.
print("Media Control:", round(media_control, 3))
print("Media Tratamiento_A:", round(media_A, 3))
print("Diferencia Control - A:", round(dif_medias, 3))
print("IC95%:", (round(li, 3), round(ls, 3)))
```

## 20. Tamaño del efecto: Cohen d

El tamaño del efecto cuantifica la magnitud estandarizada de una diferencia.

\[
d = \frac{\bar{x}_1-\bar{x}_2}{s_{pooled}}
\]

Los puntos de corte 0.2, 0.5 y 0.8 son orientativos, no leyes universales.  
La interpretación debe considerar el contexto biológico.

```python
# ============================================================
# CELDA 20. COHEN d
# ============================================================
# Calculamos la desviación estándar combinada.
sd_pooled = np.sqrt(var_pooled)

# Calculamos Cohen d.
cohen_d = dif_medias / sd_pooled

# Mostramos el resultado.
print("Cohen d =", round(cohen_d, 3))

# El signo indica dirección según el orden de la resta.
# La magnitud absoluta expresa el tamaño estandarizado del efecto.
```

## 21. Mann–Whitney U

Mann–Whitney es una alternativa basada en rangos para dos grupos independientes.

No debe describirse automáticamente como “la prueba de medianas”. Su interpretación depende de la forma de las distribuciones.

Usaremos `tiempo_respuesta_min`, una variable positiva y asimétrica.

```python
# ============================================================
# CELDA 21. VISUALIZACIÓN DEL TIEMPO DE RESPUESTA
# ============================================================
# Creamos una figura.
plt.figure(figsize=(9, 5))

# Visualizamos la distribución por grupo.
sns.boxplot(
    data=df,
    x="grupo",
    y="tiempo_respuesta_min"
)

# Añadimos puntos individuales.
sns.stripplot(
    data=df,
    x="grupo",
    y="tiempo_respuesta_min",
    alpha=0.55
)

# Etiquetas.
plt.title("Tiempo de respuesta por grupo")
plt.xlabel("Grupo")
plt.ylabel("Tiempo de respuesta (min)")

# Mostramos.
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CELDA 22. SHAPIRO-WILK PARA TIEMPO DE RESPUESTA
# ============================================================
# Evaluamos la forma de la variable dentro de cada grupo.
for grupo in df["grupo"].unique():

    # Extraemos valores.
    valores = df.loc[
        df["grupo"] == grupo,
        "tiempo_respuesta_min"
    ]

    # Ejecutamos Shapiro-Wilk.
    W, p = stats.shapiro(valores)

    # Mostramos.
    print(f"{grupo}: W = {W:.4f}, p = {p:.4f}")
```

```python
# ============================================================
# CELDA 23. MANN-WHITNEY: CONTROL VS TRATAMIENTO_A
# ============================================================
# Seleccionamos la variable de tiempo en ambos grupos.
tiempo_control = df.loc[
    df["grupo"] == "Control",
    "tiempo_respuesta_min"
]

tiempo_A = df.loc[
    df["grupo"] == "Tratamiento_A",
    "tiempo_respuesta_min"
]

# Ejecutamos una prueba bilateral.
U, p_mw = stats.mannwhitneyu(
    tiempo_control,
    tiempo_A,
    alternative="two-sided"
)

# Mostramos medianas descriptivas.
print("Mediana Control:", round(tiempo_control.median(), 2))
print("Mediana Tratamiento_A:", round(tiempo_A.median(), 2))

# Mostramos resultados inferenciales.
print("U =", round(U, 3))
print("p =", format(p_mw, ".6g"))
```

## 22. ANOVA de una vía

ANOVA evalúa la hipótesis global:

\[
H_0: \mu_C = \mu_A = \mu_B
\]

Si el resultado es significativo, sabemos que **al menos una media difiere**, pero ANOVA por sí solo no identifica qué pares son distintos.

```python
# ============================================================
# CELDA 24. ANOVA DE UNA VÍA CON STATSMODELS
# ============================================================
# Ajustamos un modelo lineal:
# longitud_cm depende del factor categórico grupo.
modelo_anova = ols(
    "longitud_cm ~ C(grupo)",
    data=df
).fit()

# Generamos la tabla ANOVA.
tabla_anova = sm.stats.anova_lm(
    modelo_anova,
    typ=2
)

# Mostramos la tabla.
display(tabla_anova)
```

## 23. Tamaño del efecto para ANOVA: η²

\[
\eta^2=\frac{SS_{entre}}{SS_{total}}
\]

Representa la proporción de variación total asociada al factor `grupo` dentro del modelo.

```python
# ============================================================
# CELDA 25. CÁLCULO DE ETA CUADRADO
# ============================================================
# Extraemos la suma de cuadrados del factor grupo.
ss_grupo = tabla_anova.loc["C(grupo)", "sum_sq"]

# Sumamos todas las sumas de cuadrados.
ss_total = tabla_anova["sum_sq"].sum()

# Calculamos eta cuadrado.
eta2 = ss_grupo / ss_total

# Mostramos.
print("Eta cuadrado (η²) =", round(eta2, 3))
```

## 24. Post hoc Tukey HSD

Si ANOVA es significativo, Tukey HSD permite identificar qué pares de grupos difieren mientras controla el error familiar en comparaciones múltiples.

```python
# ============================================================
# CELDA 26. TUKEY HSD
# ============================================================
# Aplicamos Tukey a longitud_cm entre los tres grupos.
tukey = pairwise_tukeyhsd(
    endog=df["longitud_cm"],
    groups=df["grupo"],
    alpha=0.05
)

# Mostramos la tabla de comparaciones.
print(tukey)
```

## 25. Kruskal–Wallis

Kruskal–Wallis es una alternativa no paramétrica para comparar tres o más grupos independientes.

Trabaja con rangos y evalúa una diferencia global entre distribuciones.

Usaremos `tiempo_respuesta_min`.

```python
# ============================================================
# CELDA 27. KRUSKAL-WALLIS
# ============================================================
# Extraemos tiempos de cada grupo.
t_control = df.loc[
    df["grupo"] == "Control",
    "tiempo_respuesta_min"
]

t_A = df.loc[
    df["grupo"] == "Tratamiento_A",
    "tiempo_respuesta_min"
]

t_B = df.loc[
    df["grupo"] == "Tratamiento_B",
    "tiempo_respuesta_min"
]

# Ejecutamos la prueba global.
H, p_kw = stats.kruskal(
    t_control,
    t_A,
    t_B
)

# Mostramos.
print("H =", round(H, 4))
print("p =", format(p_kw, ".6g"))
```

### Sobre el post hoc después de Kruskal–Wallis

Si Kruskal–Wallis es significativo, se requieren comparaciones múltiples basadas en rangos con corrección por multiplicidad.

Una opción habitual es **Dunn con corrección**.  
No se ejecuta aquí para mantener el taller dentro de las librerías oficiales indicadas para BIO 218.

No debe hacerse una serie de pruebas pareadas sin controlar el error por comparaciones múltiples.

## 26. Algoritmo práctico de selección

### Variable dependiente cuantitativa + grupos independientes

**Dos grupos**
- Supuestos razonables → t de Student.
- Varianzas desiguales → t de Welch.
- No normal / ordinal / rangos apropiados → Mann–Whitney.

**Tres o más grupos**
- Supuestos razonables → ANOVA.
- No paramétrica / ordinal → Kruskal–Wallis.

Después de una prueba global significativa:
- ANOVA → Tukey HSD si corresponde.
- Kruskal–Wallis → procedimiento post hoc basado en rangos con ajuste múltiple.

### Regla pedagógica

**Pregunta → diseño → variable → grupos → supuestos → prueba → efecto → interpretación.**
